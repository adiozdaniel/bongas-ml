# 🧠 The Swahili Brain: Supervised Fine-Tuning (SFT)

This notebook orchestrates the Golden SFT for the **Sovereign Language Model (SLM)**. It fine-tunes a base transformer on technical Swahili and Sheng datasets using **LoRA (Low-Rank Adaptation)**.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments, Trainer
from peft import LoraConfig, get_peft_model
from datasets import load_dataset
import torch

# 1. Setup Base Model & LoRA
model_id = "Qwen/Qwen1.5-0.5B"
peft_config = LoraConfig(
    r=8, lora_alpha=32, target_modules=["q_proj", "v_proj"], 
    lora_dropout=0.05, task_type="CAUSAL_LM"
)

model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.float16)
model = get_peft_model(model, peft_config)

# 2. Load Technical Swahili/Sheng Dataset
# dataset = load_dataset("json", data_files="swahili_technical.json")

# 3. Execute Training
trainer = Trainer(
    model=model,
    args=TrainingArguments(output_dir="./sft_output", per_device_train_batch_size=4, learning_rate=2e-4)
)
# trainer.train()

print("SFT Pipeline Ready. Exporting to Sovereign Registry...")
```
